```python
# Complete NEAT trainer
class CompleteNEATTrainer:
    """Complete NEAT trainer with all functionalities"""
    
    def __init__(self, config=None):
        # Default configuration
        if config is None:
            config = {
                'generations': 100,
                'population_size': 100,
                'episodes_per_individual': 1,
                'max_steps': 300,
                'use_gpu': ENV_INFO['gpu_available'],
                'save_checkpoint_interval': 10,
                'target_fitness': 5.0,
                'enable_visualization': True,
                'enable_analysis': True
            }
        
        self.config = config
        
        print("🚀 Complete NEAT trainer started")
        print("=" * 60)
        print(f"📊 Training Configuration:")
        for key, value in self.config.items():
            print(f"  - {key}: {value}")
        print("=" * 60)
        
        # Initialize NEAT configuration
        self.neat_config = NEATConfig()
        self.neat_config.population_size = self.config['population_size']
        
        # Set optimized mutation parameters
        self.neat_config.weight_mutation_rate = 0.8
        self.neat_config.weight_mutation_power = 0.1
        self.neat_config.add_connection_rate = 0.1
        self.neat_config.add_node_rate = 0.05
        self.neat_config.species_threshold = 3.0
        
        # Initialize EvoJAX compatible SlimeVolley environment
        self.slimevolley_neat = NEATEvoJAXInterface(
            max_steps=self.config['max_steps'],
            test=False,  # Training mode
            use_jax=self.config['use_gpu']
        )
        
        # Initialize NEAT population
        self.population = NEATPopulation(
            input_size=self.slimevolley_neat.input_size,
            output_size=self.slimevolley_neat.output_size,
            population_size=self.config['population_size']
        )
        
        # Set NEAT parameters
        self.population.species_threshold = self.neat_config.species_threshold
        self.population.species_elite_size = self.neat_config.species_elite_size
        self.population.species_stagnation_limit = self.neat_config.species_stagnation_limit
        self.population.elite_size = self.neat_config.elite_size
        
        # Initialize analyzers
        if self.config['enable_analysis']:
            self.evolution_analyzer = NetworkEvolutionAnalyzer()
            self.activation_manager = ActivationFunctionManager()
        
        # Training statistics
        self.training_stats = {
            'start_time': time.time(),
            'generations': [],
            'best_fitness_history': [],
            'avg_fitness_history': [],
            'species_count_history': [],
            'total_time': 0,
            'best_genome': None
        }
        
        print("✅ Trainer initialization completed")
        print(f"📊 Environment Information: {self.slimevolley_neat.get_environment_info()}")
```

```python
# Add trainer methods
def add_trainer_methods():
    """Add methods to trainer"""
    
    # Population evaluation method
    def evaluate_population(self, population):
        """Evaluate population"""
        print(f"🔍 Evaluating population...")
        
        genomes = population.population
        
        # Use EvoJAX compatibility interface for batch evaluation
        if self.config['use_gpu']:
            try:
                print("🚀 Using EvoJAX compatibility interface for batch evaluation...")
                fitnesses = self.slimevolley_neat.evaluate_population_batch(
                    genomes,
                    num_episodes=self.config['episodes_per_individual'],
                    batch_size=32  # Optimized batch size
                )
            except Exception as e:
                print(f"⚠️  EvoJAX batch evaluation failed, falling back to serial evaluation: {e}")
                fitnesses = self._evaluate_serial(genomes)
        else:
            fitnesses = self._evaluate_serial(genomes)
        
        # Set fitness values
        for genome, fitness in zip(genomes, fitnesses):
            genome.fitness = fitness
            
        return fitnesses
    
    # Serial evaluation method
    def _evaluate_serial(self, genomes):
        """Serial evaluation"""
        fitnesses = []
        for i, genome in enumerate(genomes):
            if i % 10 == 0:
                print(f"  Evaluating individual {i+1}/{len(genomes)}")
            fitness = self.slimevolley_neat.evaluate_genome(genome)
            fitnesses.append(fitness)
        return fitnesses
    
    # Train generation method
    def train_generation(self, generation_num):
        """Train one generation"""
        print(f"\n🔄 Generation {generation_num} training started")
        gen_start_time = time.time()
        
        # Evaluate current population
        fitnesses = self.evaluate_population(self.population)
        
        # Collect statistics
        best_fitness = max(fitnesses)
        avg_fitness = sum(fitnesses) / len(fitnesses)
        species_count = len(self.population.species)
        
        # Update best genome
        best_genome_idx = fitnesses.index(best_fitness)
        self.training_stats['best_genome'] = self.population.population[best_genome_idx]
        
        # Display current generation results
        print(f"📊 Generation {generation_num} Results:")
        print(f"  - Best Fitness: {best_fitness:.3f}")
        print(f"  - Average Fitness: {avg_fitness:.3f}")
        print(f"  - Species Count: {species_count}")
        
        # Record statistics
        self.training_stats['generations'].append({
            'generation': generation_num,
            'best_fitness': best_fitness,
            'avg_fitness': avg_fitness,
            'species_count': species_count,
            'time': time.time() - gen_start_time
        })
        
        self.training_stats['best_fitness_history'].append(best_fitness)
        self.training_stats['avg_fitness_history'].append(avg_fitness)
        self.training_stats['species_count_history'].append(species_count)
        
        # Analyze evolution process
        if self.config['enable_analysis']:
            self.evolution_analyzer.add_generation(
                self.population.population, generation_num
            )
        
        # Check if target reached
        if best_fitness >= self.config['target_fitness']:
            print(f"🎯 Target achieved! Best fitness: {best_fitness:.3f}")
            return True
        
        # Evolve to next generation
        if generation_num < self.config['generations']:
            print(f"🔄 Evolving to generation {generation_num + 1}...")
            new_population = self.population.reproduce()
            if isinstance(new_population, list):
                self.population.population = new_population
            else:
                self.population = new_population
        
        gen_time = time.time() - gen_start_time
        print(f"⏱️  Generation {generation_num} took: {gen_time:.1f} seconds")
        
        return False
    
    # Add methods to class
    CompleteNEATTrainer.evaluate_population = evaluate_population
    CompleteNEATTrainer._evaluate_serial = _evaluate_serial
    CompleteNEATTrainer.train_generation = train_generation

# Add methods
add_trainer_methods()
print("✅ Trainer methods added successfully")
```

## 🔄 Backprop NEAT Implementation

Hybrid approach: NEAT for architecture evolution + Backpropagation for weight optimization

```python
# Complete Backprop NEAT Implementation
import jax
import jax.numpy as jnp
import optax
from typing import Dict, List, Tuple, Optional
import numpy as np

class BackpropNEATTrainer:
    """Complete Hybrid NEAT + Backpropagation trainer"""
    
    def __init__(self, learning_rate=0.01, epochs_per_genome=10):
        self.learning_rate = learning_rate
        self.epochs_per_genome = epochs_per_genome
        self.optimizer = optax.adam(learning_rate)
        
    def _get_topological_order(self, genome):
        """Get nodes in topological order for feed-forward computation"""
        nodes = genome.nodes
        connections = {k: v for k, v in genome.connections.items() if v.enabled}
        
        # Build adjacency list and in-degree count
        adj_list = {node_id: [] for node_id in nodes.keys()}
        in_degree = {node_id: 0 for node_id in nodes.keys()}
        
        for conn in connections.values():
            adj_list[conn.input_node].append(conn.output_node)
            in_degree[conn.output_node] += 1
        
        # Topological sort using Kahn's algorithm
        queue = [node_id for node_id in nodes.keys() if in_degree[node_id] == 0]
        result = []
        
        while queue:
            node = queue.pop(0)
            result.append(node)
            
            for neighbor in adj_list[node]:
                in_degree[neighbor] -= 1
                if in_degree[neighbor] == 0:
                    queue.append(neighbor)
        
        return result
    
    def create_trainable_network(self, genome):
        """Convert NEAT genome to trainable JAX network parameters"""
        # Extract network structure from genome
        nodes = genome.nodes
        connections = genome.connections
        
        # Create trainable parameters
        params = {}
        for conn_id, conn in connections.items():
            if conn.enabled:
                weight_key = f"weight_{conn.input_node}_{conn.output_node}"
                params[weight_key] = jnp.array(conn.weight, dtype=jnp.float32)
        
        return params
    
    def forward_pass(self, params, inputs, genome):
        """JAX-based forward pass through evolved architecture"""
        # Convert genome topology to executable JAX computation
        node_values = {}
        
        # Initialize input nodes
        input_nodes = [nid for nid, node in genome.nodes.items() 
                       if node.node_type.value == 'input']
        for i, input_val in enumerate(inputs):
            if i < len(input_nodes):
                node_values[input_nodes[i]] = input_val
        
        # Process nodes in topological order
        for node_id in self._get_topological_order(genome):
            if node_id not in node_values:  # Skip input nodes
                # Compute weighted sum of inputs
                weighted_sum = 0.0
                for conn_id, conn in genome.connections.items():
                    if conn.output_node == node_id and conn.enabled:
                        weight_key = f"weight_{conn.input_node}_{conn.output_node}"
                        if weight_key in params and conn.input_node in node_values:
                            weighted_sum += params[weight_key] * node_values[conn.input_node]
                
                # Apply activation function (differentiable only)
                node = genome.nodes[node_id]
                if node.activation.value == 'relu':
                    node_values[node_id] = jnp.maximum(0, weighted_sum)
                elif node.activation.value == 'tanh':
                    node_values[node_id] = jnp.tanh(weighted_sum)
                elif node.activation.value == 'sigmoid':
                    node_values[node_id] = jax.nn.sigmoid(weighted_sum)
                else:
                    node_values[node_id] = weighted_sum  # Linear
        
        # Extract outputs
        outputs = []
        output_nodes = [nid for nid, node in genome.nodes.items() 
                        if node.node_type.value == 'output']
        for node_id in output_nodes:
            if node_id in node_values:
                outputs.append(node_values[node_id])
        
        return jnp.array(outputs)
    
    def compute_loss(self, params, genome, batch_data, batch_labels):
        """Compute loss for a batch of data"""
        batch_size = len(batch_data)
        total_loss = 0.0
        
        for i in range(batch_size):
            inputs = batch_data[i]
            target = batch_labels[i]
            
            # Forward pass
            prediction = self.forward_pass(params, inputs, genome)
            
            # Compute loss (binary cross-entropy for classification)
            if len(prediction) == 1:  # Binary classification
                pred_prob = jax.nn.sigmoid(prediction[0])
                loss = -target * jnp.log(pred_prob + 1e-8) - (1 - target) * jnp.log(1 - pred_prob + 1e-8)
            else:  # Multi-class classification
                pred_probs = jax.nn.softmax(prediction)
                target_one_hot = jax.nn.one_hot(target, len(prediction))
                loss = -jnp.sum(target_one_hot * jnp.log(pred_probs + 1e-8))
            
            total_loss += loss
        
        return total_loss / batch_size
    
    def train_weights(self, genome, training_data, training_labels, epochs=None, batch_size=32):
        """Complete weight training using backpropagation"""
        if epochs is None:
            epochs = self.epochs_per_genome
        
        # Initialize parameters
        params = self.create_trainable_network(genome)
        opt_state = self.optimizer.init(params)
        
        # Training loop
        n_samples = len(training_data)
        
        for epoch in range(epochs):
            epoch_loss = 0.0
            
            # Process data in batches
            for start_idx in range(0, n_samples, batch_size):
                end_idx = min(start_idx + batch_size, n_samples)
                batch_data = training_data[start_idx:end_idx]
                batch_labels = training_labels[start_idx:end_idx]
                
                # Compute loss and gradients
                loss, grads = jax.value_and_grad(self.compute_loss)(params, genome, batch_data, batch_labels)
                
                # Update parameters
                updates, opt_state = self.optimizer.update(grads, opt_state)
                params = optax.apply_updates(params, updates)
                
                epoch_loss += loss
            
            # Print progress
            if epoch % 5 == 0:
                avg_loss = epoch_loss / (n_samples // batch_size + 1)
                print(f"  Epoch {epoch+1}/{epochs}, Loss: {avg_loss:.4f}")
        
        return params
    
    def evaluate_genome(self, genome, test_data, test_labels):
        """Evaluate trained genome on test data"""
        params = self.create_trainable_network(genome)
        
        correct = 0
        total = len(test_data)
        
        for i in range(total):
            inputs = test_data[i]
            target = test_labels[i]
            
            prediction = self.forward_pass(params, inputs, genome)
            
            if len(prediction) == 1:  # Binary classification
                pred_class = 1 if prediction[0] > 0 else 0
            else:  # Multi-class classification
                pred_class = jnp.argmax(prediction)
            
            if pred_class == target:
                correct += 1
        
        accuracy = correct / total
        return accuracy

# Test the implementation
print("✅ Complete Backprop NEAT implementation added")
print("  - Forward pass: Fully implemented with JAX")
print("  - Weight training: Complete backpropagation with Optax")
print("  - Loss computation: Binary and multi-class classification")
print("  - Evaluation: Accuracy computation for test data")
```

```python
# Generate Classification Datasets
import numpy as np
import matplotlib.pyplot as plt

def generate_circle_dataset(n_samples=1000, noise=0.1):
    """Generate circle classification dataset"""
    # Generate points in a circle
    theta = np.random.uniform(0, 2*np.pi, n_samples)
    r = np.random.uniform(0.3, 0.7, n_samples)
    
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    
    # Add noise
    x += np.random.normal(0, noise, n_samples)
    y += np.random.normal(0, noise, n_samples)
    
    # Create labels (inside/outside circle)
    labels = (x**2 + y**2 < 0.5**2).astype(int)
    
    return np.column_stack([x, y]), labels

def generate_spiral_dataset(n_samples=1000, noise=0.1):
    """Generate spiral classification dataset"""
    # Generate spiral data
    n = n_samples // 2
    
    # First spiral
    theta1 = np.linspace(0, 4*np.pi, n)
    r1 = np.linspace(0, 2, n)
    x1 = r1 * np.cos(theta1)
    y1 = r1 * np.sin(theta1)
    
    # Second spiral
    theta2 = np.linspace(0, 4*np.pi, n)
    r2 = np.linspace(0, 2, n)
    x2 = r2 * np.cos(theta2 + np.pi)
    y2 = r2 * np.sin(theta2 + np.pi)
    
    # Combine and add noise
    x = np.concatenate([x1, x2]) + np.random.normal(0, noise, n_samples)
    y = np.concatenate([y1, y2]) + np.random.normal(0, noise, n_samples)
    labels = np.concatenate([np.zeros(n), np.ones(n)])
    
    return np.column_stack([x, y]), labels

def generate_xor_dataset(n_samples=1000, noise=0.1):
    """Generate XOR classification dataset"""
    # Generate XOR pattern
    x = np.random.uniform(-1, 1, n_samples)
    y = np.random.uniform(-1, 1, n_samples)
    
    # XOR logic: (x > 0) XOR (y > 0)
    labels = ((x > 0) != (y > 0)).astype(int)
    
    # Add noise
    x += np.random.normal(0, noise, n_samples)
    y += np.random.normal(0, noise, n_samples)
    
    return np.column_stack([x, y]), labels

# Generate all datasets
circle_data, circle_labels = generate_circle_dataset()
spiral_data, spiral_labels = generate_spiral_dataset()
xor_data, xor_labels = generate_xor_dataset()

print("✅ Generated classification datasets:")
print(f"  - Circle: {len(circle_data)} samples")
print(f"  - Spiral: {len(spiral_data)} samples")
print(f"  - XOR: {len(xor_data)} samples")
```

```python
# JAX-Native Network Execution and Gradient Computation
import jax
import jax.numpy as jnp
from jax import jit, grad, vmap
import optax

class JAXNativeNEAT:
    """JAX-native NEAT implementation with full differentiability"""
    
    def __init__(self):
        # JAX-native activation functions
        self.activations = {
            'relu': jnp.maximum,
            'tanh': jnp.tanh,
            'sigmoid': jax.nn.sigmoid,
            'leaky_relu': lambda x: jnp.where(x >= 0, x, 0.01 * x),
            'elu': lambda x: jnp.where(x >= 0, x, jnp.exp(x) - 1),
            'swish': lambda x: x * jax.nn.sigmoid(x)
        }
        
        # JIT-compiled forward pass
        self.forward_pass_jitted = jit(self._forward_pass_impl)
        
    def _forward_pass_impl(self, params, inputs, genome_info):
        """JAX-native forward pass implementation"""
        # Extract genome structure
        nodes = genome_info['nodes']
        connections = genome_info['connections']
        
        # Initialize node values
        node_values = {}
        
        # Set input values
        for i, input_val in enumerate(inputs):
            node_values[i] = input_val
        
        # Process nodes in topological order
        for node_id in genome_info['topological_order']:
            if node_id not in node_values:  # Skip input nodes
                # Compute weighted sum
                weighted_sum = 0.0
                for conn in connections:
                    if conn['output_node'] == node_id and conn['enabled']:
                        weight_key = f"weight_{conn['input_node']}_{conn['output_node']}"
                        if weight_key in params and conn['input_node'] in node_values:
                            weighted_sum += params[weight_key] * node_values[conn['input_node']]
                
                # Apply activation function
                node = nodes[node_id]
                activation_name = node['activation']
                if activation_name in self.activations:
                    node_values[node_id] = self.activations[activation_name](weighted_sum)
                else:
                    node_values[node_id] = weighted_sum  # Linear
        
        # Extract outputs
        outputs = []
        for node_id, node in nodes.items():
            if node['node_type'] == 'output' and node_id in node_values:
                outputs.append(node_values[node_id])
        
        return jnp.array(outputs)
    
    def create_trainable_network(self, genome):
        """Convert genome to JAX trainable format"""
        # Extract network structure
        nodes = {}
        connections = []
        
        for node_id, node in genome.nodes.items():
            nodes[node_id] = {
                'node_type': node.node_type.value,
                'activation': node.activation.value
            }
        
        for conn in genome.connections.values():
            if conn.enabled:
                connections.append({
                    'input_node': conn.input_node,
                    'output_node': conn.output_node,
                    'enabled': conn.enabled
                })
        
        # Create trainable parameters
        params = {}
        for conn in connections:
            weight_key = f"weight_{conn['input_node']}_{conn['output_node']}"
            params[weight_key] = jnp.array(genome.connections[conn['input_node']].weight, dtype=jnp.float32)
        
        # Compute topological order
        topological_order = self._compute_topological_order(nodes, connections)
        
        genome_info = {
            'nodes': nodes,
            'connections': connections,
            'topological_order': topological_order
        }
        
        return params, genome_info
    
    def _compute_topological_order(self, nodes, connections):
        """Compute topological order for JAX execution"""
        # Build adjacency list and in-degree count
        adj_list = {node_id: [] for node_id in nodes.keys()}
        in_degree = {node_id: 0 for node_id in nodes.keys()}
        
        for conn in connections:
            adj_list[conn['input_node']].append(conn['output_node'])
            in_degree[conn['output_node']] += 1
        
        # Topological sort
        queue = [node_id for node_id in nodes.keys() if in_degree[node_id] == 0]
        result = []
        
        while queue:
            node = queue.pop(0)
            result.append(node)
            
            for neighbor in adj_list[node]:
                in_degree[neighbor] -= 1
                if in_degree[neighbor] == 0:
                    queue.append(neighbor)
        
        return result
    
    def compute_gradients(self, params, genome_info, data, labels):
        """Compute gradients using JAX automatic differentiation"""
        def loss_fn(params):
            total_loss = 0.0
            for i in range(len(data)):
                prediction = self.forward_pass_jitted(params, data[i], genome_info)
                target = labels[i]
                
                if len(prediction) == 1:  # Binary classification
                    pred_prob = jax.nn.sigmoid(prediction[0])
                    loss = -target * jnp.log(pred_prob + 1e-8) - (1 - target) * jnp.log(1 - pred_prob + 1e-8)
                else:  # Multi-class
                    pred_probs = jax.nn.softmax(prediction)
                    target_one_hot = jax.nn.one_hot(target, len(prediction))
                    loss = -jnp.sum(target_one_hot * jnp.log(pred_probs + 1e-8))
                
                total_loss += loss
            
            return total_loss / len(data)
        
        # Compute gradients
        grads = grad(loss_fn)(params)
        return grads
    
    def train_network(self, params, genome_info, data, labels, learning_rate=0.01, epochs=10):
        """Train network using JAX gradients"""
        optimizer = optax.adam(learning_rate)
        opt_state = optimizer.init(params)
        
        for epoch in range(epochs):
            # Compute gradients
            grads = self.compute_gradients(params, genome_info, data, labels)
            
            # Update parameters
            updates, opt_state = optimizer.update(grads, opt_state)
            params = optax.apply_updates(params, updates)
            
            # Compute loss for monitoring
            if epoch % 5 == 0:
                loss = self._compute_loss(params, genome_info, data, labels)
                print(f"  Epoch {epoch+1}/{epochs}, Loss: {loss:.4f}")
        
        return params
    
    def _compute_loss(self, params, genome_info, data, labels):
        """Compute loss for monitoring"""
        total_loss = 0.0
        for i in range(len(data)):
            prediction = self.forward_pass_jitted(params, data[i], genome_info)
            target = labels[i]
            
            if len(prediction) == 1:  # Binary classification
                pred_prob = jax.nn.sigmoid(prediction[0])
                loss = -target * jnp.log(pred_prob + 1e-8) - (1 - target) * jnp.log(1 - pred_prob + 1e-8)
            else:  # Multi-class
                pred_probs = jax.nn.softmax(prediction)
                target_one_hot = jax.nn.one_hot(target, len(prediction))
                loss = -jnp.sum(target_one_hot * jnp.log(pred_probs + 1e-8))
            
            total_loss += loss
        
        return total_loss / len(data)

# Test JAX integration
print("✅ JAX-native integration implemented:")
print("  - Fully differentiable activation functions")
print("  - JIT-compiled forward pass")
print("  - Automatic gradient computation")
print("  - JAX-native network training")
```

```python
# GIF Generation for SlimeVolley Evolution
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from PIL import Image
import io
import base64
from datetime import datetime
import numpy as np

class SlimeVolleyGIFGenerator:
    """Generate animated GIFs showing NEAT evolution in SlimeVolley"""
    
    def __init__(self):
        self.frames = []
        self.evolution_data = []
        
    def record_evolution_frame(self, generation, best_genome, population_stats):
        """Record a frame of evolution progress"""
        frame_data = {
            'generation': generation,
            'best_genome': best_genome,
            'population_stats': population_stats,
            'timestamp': datetime.now()
        }
        self.evolution_data.append(frame_data)
        
        # Create visualization frame
        frame = self._create_evolution_frame(frame_data)
        self.frames.append(frame)
        
        return frame
    
    def _create_evolution_frame(self, frame_data):
        """Create a single frame showing evolution progress"""
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 10))
        
        gen = frame_data['generation']
        stats = frame_data['population_stats']
        
        # 1. Fitness distribution
        if 'fitness_distribution' in stats:
            ax1.hist(stats['fitness_distribution'], bins=20, alpha=0.7, color='skyblue', edgecolor='black')
            ax1.set_title(f'Generation {gen}: Fitness Distribution')
            ax1.set_xlabel('Fitness')
            ax1.set_ylabel('Count')
            ax1.grid(True, alpha=0.3)
        
        # 2. Best genome network visualization
        if 'best_genome' in frame_data:
            self._visualize_genome_network(ax2, frame_data['best_genome'], f'Best Genome (Gen {gen})')
        
        # 3. Evolution metrics
        if 'evolution_metrics' in stats:
            metrics = stats['evolution_metrics']
            x_pos = np.arange(len(metrics))
            ax3.bar(x_pos, list(metrics.values()), color=['red', 'blue', 'green', 'orange'])
            ax3.set_title(f'Generation {gen}: Evolution Metrics')
            ax3.set_ylabel('Value')
            ax3.set_xticks(x_pos)
            ax3.set_xticklabels(list(metrics.keys()), rotation=45)
            ax3.grid(True, alpha=0.3)
        
        # 4. Progress over time
        if len(self.evolution_data) > 1:
            generations = [d['generation'] for d in self.evolution_data]
            best_fitness = [d['population_stats'].get('best_fitness', 0) for d in self.evolution_data]
            avg_fitness = [d['population_stats'].get('avg_fitness', 0) for d in self.evolution_data]
            
            ax4.plot(generations, best_fitness, 'b-', label='Best Fitness', linewidth=2)
            ax4.plot(generations, avg_fitness, 'r--', label='Average Fitness', alpha=0.7)
            ax4.set_title('Evolution Progress')
            ax4.set_xlabel('Generation')
            ax4.set_ylabel('Fitness')
            ax4.legend()
            ax4.grid(True, alpha=0.3)
        
        plt.tight_layout()
        
        # Convert to image
        buf = io.BytesIO()
        plt.savefig(buf, format='png', dpi=100, bbox_inches='tight')
        buf.seek(0)
        img = Image.open(buf)
        plt.close()
        
        return img
    
    def _visualize_genome_network(self, ax, genome, title):
        """Visualize a genome's network structure"""
        try:
            import networkx as nx
            
            # Create network graph
            G = nx.DiGraph()
            
            # Add nodes
            for node_id, node in genome.nodes.items():
                G.add_node(node_id, node_type=node.node_type.value, activation=node.activation.value)
            
            # Add edges
            for conn in genome.connections.values():
                if conn.enabled:
                    G.add_edge(conn.input_node, conn.output_node, weight=conn.weight)
            
            # Layout
            pos = nx.spring_layout(G, k=1, iterations=50)
            
            # Draw nodes
            node_colors = []
            for node in G.nodes():
                node_type = G.nodes[node]['node_type']
                if node_type == 'input':
                    node_colors.append('red')
                elif node_type == 'output':
                    node_colors.append('green')
                else:
                    node_colors.append('blue')
            
            nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=300, ax=ax)
            nx.draw_networkx_edges(G, pos, edge_color='gray', arrows=True, arrowsize=10, ax=ax)
            nx.draw_networkx_labels(G, pos, font_size=8, ax=ax)
            
            ax.set_title(title)
            ax.axis('off')
            
        except ImportError:
            ax.text(0.5, 0.5, 'NetworkX not available\nfor network visualization', 
                   ha='center', va='center', transform=ax.transAxes)
            ax.set_title(title)
    
    def generate_evolution_gif(self, filename='slimevolley_evolution.gif', fps=2):
        """Generate animated GIF from recorded frames"""
        if not self.frames:
            print("❌ No frames recorded")
            return None
        
        print(f"🎬 Generating evolution GIF with {len(self.frames)} frames...")
        
        # Save first frame
        self.frames[0].save(filename, save_all=True, append_images=self.frames[1:], 
                           duration=1000//fps, loop=0)
        
        print(f"✅ Evolution GIF saved as {filename}")
        return filename
    
    def create_training_animation(self, fitness_history, species_history, 
                                   save_path='training_evolution.gif'):
        """Create animated GIF showing training evolution"""
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
        
        def animate(frame):
            ax1.clear()
            ax2.clear()
            
            # Plot fitness up to current frame
            if frame < len(fitness_history):
                ax1.plot(fitness_history[:frame+1], 'b-', linewidth=2)
                ax1.set_title(f'Fitness Evolution (Gen {frame+1})')
                ax1.set_xlabel('Generation')
                ax1.set_ylabel('Best Fitness')
                ax1.grid(True, alpha=0.3)
                ax1.set_ylim(0, max(fitness_history) * 1.1)
            
            # Plot species count up to current frame
            if frame < len(species_history):
                ax2.plot(species_history[:frame+1], 'r-', linewidth=2)
                ax2.set_title(f'Species Count (Gen {frame+1})')
                ax2.set_xlabel('Generation')
                ax2.set_ylabel('Species Count')
                ax2.grid(True, alpha=0.3)
                ax2.set_ylim(0, max(species_history) * 1.1)
            
            plt.tight_layout()
        
        # Create animation
        anim = animation.FuncAnimation(fig, animate, frames=min(len(fitness_history), len(species_history)), 
                                       interval=500, repeat=True)
        
        # Save as GIF
        anim.save(save_path, writer='pillow', fps=2)
        print(f"✅ Training animation saved as {save_path}")
        
        return anim

# Test GIF generation
print("✅ SlimeVolley GIF generation implemented:")
print("  - Evolution frame recording")
print("  - Network structure visualization")
print("  - Training progress animation")
print("  - GIF export functionality")
```

```python
# Network Architecture Visualization and Analysis
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np
from typing import Dict, List, Tuple

class NetworkVisualizer:
    """Advanced network visualization and analysis"""
    
    def __init__(self):
        self.colors = {
            'input': '#ff7f7f',      # Red
            'hidden': '#7f7fff',     # Blue
            'output': '#7fff7f'      # Green
        }
    
    def visualize_genome(self, genome, title="Network Architecture", figsize=(12, 8)):
        """Visualize NEAT genome as a directed graph"""
        G = nx.DiGraph()
        
        # Add nodes
        for node_id, node in genome.nodes.items():
            G.add_node(node_id, 
                       type=node.node_type.value,
                       activation=node.activation.value,
                       pos=(node.x_position, node.y_position))
        
        # Add connections
        for conn_id, conn in genome.connections.items():
            if conn.enabled:
                G.add_edge(conn.input_node, conn.output_node, 
                           weight=conn.weight,
                           innovation=conn.innovation_number)
        
        # Create visualization
        plt.figure(figsize=figsize)
        pos = nx.spring_layout(G, k=1, iterations=50)
        
        # Draw nodes with different colors
        node_colors = [self.colors[G.nodes[node]['type']] for node in G.nodes()]
        nx.draw_networkx_nodes(G, pos, node_color=node_colors, 
                               node_size=500, alpha=0.8)
        
        # Draw edges with weight-based thickness
        edge_weights = [abs(G[u][v]['weight']) for u, v in G.edges()]
        edge_widths = [1 + 2 * w for w in edge_weights]
        nx.draw_networkx_edges(G, pos, width=edge_widths, 
                               edge_color='gray', alpha=0.6, 
                               arrows=True, arrowsize=20)
        
        # Add labels
        labels = {node: f"{node}\n{G.nodes[node]['activation']}" 
                 for node in G.nodes()}
        nx.draw_networkx_labels(G, pos, labels, font_size=8)
        
        plt.title(title)
        plt.axis('off')
        plt.tight_layout()
        plt.show()
        
        return G
    
    def analyze_complexity(self, genome):
        """Analyze network complexity metrics"""
        nodes = genome.nodes
        connections = genome.connections
        
        # Count node types
        input_count = sum(1 for n in nodes.values() if n.node_type.value == 'input')
        hidden_count = sum(1 for n in nodes.values() if n.node_type.value == 'hidden')
        output_count = sum(1 for n in nodes.values() if n.node_type.value == 'output')
        
        # Count connections
        enabled_connections = sum(1 for c in connections.values() if c.enabled)
        total_connections = len(connections)
        
        # Calculate complexity metrics
        complexity_score = (hidden_count * 2 + enabled_connections) / (input_count + output_count)
        
        # Activation function distribution
        activation_counts = {}
        for node in nodes.values():
            act = node.activation.value
            activation_counts[act] = activation_counts.get(act, 0) + 1
        
        return {
            'input_nodes': input_count,
            'hidden_nodes': hidden_count,
            'output_nodes': output_count,
            'enabled_connections': enabled_connections,
            'total_connections': total_connections,
            'complexity_score': complexity_score,
            'activation_distribution': activation_counts
        }
    
    def plot_complexity_evolution(self, complexity_history):
        """Plot complexity evolution over generations"""
        generations = list(range(len(complexity_history)))
        scores = [h['complexity_score'] for h in complexity_history]
        
        plt.figure(figsize=(10, 6))
        plt.plot(generations, scores, 'b-', linewidth=2, marker='o')
        plt.xlabel('Generation')
        plt.ylabel('Complexity Score')
        plt.title('Network Complexity Evolution')
        plt.grid(True, alpha=0.3)
        plt.show()

print("✅ Network visualization and analysis added")
```
